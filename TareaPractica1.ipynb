### **1. Coordenadas polares y cartesianas:** 
#### Supongamos que se nos da (es decir, el usuario introduce) la posición de un punto en un espacio bidimensional:

#### **a)** Si el punto se especifica  en coordenadas polares $r$,$\theta$. Escribe un programa (en `python` sin usar la biblioteca `numpy`) que calcule e imprima la conversión a coordenadas cartesianas $x$, $y$.

```python
r = float(input("introduce el valor del radio: "))

θ_grados = float(input("Ingresa el ángulo θ (en grados): "))

# Convertimos del ángulo de grados a radianes

import math as mt # para usar la función radianes
θ_rad = mt.radians(θ_grados)

# Calculamos las componentes cartesianas x e y
x = r * mt.cos(θ_rad)
y = r * mt.sin(θ_rad)

print("El valor de x es: ", x)
print("El valor de y es: ", y)
```

#### **b)** Si el punto se especifica en coordenadas cartesianas $x$, $y$. Escribe un programa (en `python` sin usar `numpy`) que calcule e imprima la conversión a coordenadas polares $r$,$\theta$.

```python
x = float(input("Ingresa la coordenada x: "))
y = float(input("Ingresa la coordenada y: "))

# Cálculo del radio r
r = mt.sqrt(x**2 + y**2)

# Cálculo del ángulo θ en radianes y su conversión a grados
θ_rad = mt.atan2(y, x)
θ_grados = mt.degrees(θ_rad)

print("El valor del radio es: ", r)
print("El valor del ángulo en grados es: ", θ_grados)
```

### **2. Caída de una pelota desde una torre:** 
#### Se deja caer una pelota desde una torre de altura $h$. Tiene velocidad inicial cero y acelera hacia abajo con la gravedad.

Escribe un programa que le pida al usuario que ingrese la altura en metros de la torre y luego calcule e imprima el tiempo en segundos hasta que la pelota toque el suelo (ignora la resistencia del aire).

Después, usa tu programa para calcular el tiempo de una pelota lanzada desde una torre de $100m$ de altura.\\


```python
h = float(input("Escriba la altura en metros:" ))

#Definimos la aceleracion de la gravedad como:

g = 9.81
```

El movimiento vertical de caída libre bajo la influencia de la aceleración gravitatoria ($g$) se modela con la ecuación:
$$y(t) = h - \frac{1}{2}gt^2$$ 
Para calcular el tiempo ($t$) que tarda en tocar el suelo ($y(t) = 0$), despejamos la variable de tiempo de la ecuación:
$$0 = h - \frac{1}{2}gt^2 \implies h = \frac{1}{2}gt^2 \implies t^2 = \frac{2h}{g} \implies t = \sqrt{\frac{2h}{g}}$$
Las unidades de $t$ ya están dadas en segundos, ahora sólo hay que definir la ecuación, utilizar la altura dada e imprimir el resultado.

```python
import math as mt #para usar la funcion raiz cuadrada

t = mt.sqrt((2*h)/g)
print("El tiempo que tarda la pelota en caer (en segundos) es:", t)
```

Si ahora la altura es especificamente 100 metros:

```python
h = 100
t = mt.sqrt((2*h)/g)
print("El tiempo que tarda la pelota en caer (en segundos) es:", t)
```

### **4. Paso de potencial cuántico.** 
#### Un conocido problema de mecánica cuántica involucra una partícula de masa $m$ que encuentra un paso de potencial unidimensional, como este:

<center><img src="qstep2.png" width="50%" class="center"/></center>

#### La partícula con energía cinética inicial $E$ y vector de onda $k_1=\sqrt{2mE}/\hbar$ entra por la izquierda y encuentra un salto repentino en la energía potencial de altura $V$ en la posición $x=0$. Resolviendo la ecuación de Schrödinger, se puede demostrar que cuando $E>V$ la partícula puede 

#### (a) **pasar el escalón**, en cuyo caso tiene una energía cinética menor a $E-V$ en el otro lado y un vector de onda correspondientemente menor a $k_2=\sqrt{2m(E-V)}/\hbar$, \; o bien (b) **reflejarse**, manteniendo toda su energía cinética y un vector de onda sin cambios, pero moviéndose en la dirección opuesta. Las probabilidades $T$ y $R$ de transmisión y reflexión están dadas por:

<h3 align="center">$T = {4k_1k_2\over(k_1+k_2)^2}\,,\qquad$
$R = \biggl( {k_1-k_2\over k_1+k_2} \biggr)^2.$</h3>

#### Supongamos que tenemos una partícula con una masa igual a la masa del electrón. $m=9.11\times10^{-31}\,$kg y energía de $10\,$eV, al encontrar un escalón de potencial de altura $9\,$eV. Escribe un programa en \texttt{Python} para calcular e imprimir las probabilidades de transmisión y reflexión utilizando las fórmulas anteriores.


```python
'''Dado que del enunciado ya conocemos las ecuaciones y los parametros requeridos lo único que nos queda hacer es pasarlo a Python'''
```

```python
import math as mt

# Definimos las Constantes Físicas en unidades del S.I.
m = 9.11e-31          
hbar = 1.0545718e-34  # Constante de Planck reducida (J * s)
eV_a_J = 1.60217663e-19  # Factor de conversión de eV a Julios

# Condiciones iniciales dadas por el problema (en eV)
E_eV = 10.0
V_eV = 9.0

# Conversión de las energías a Julios (J)
E = E_eV * eV_a_J
V = V_eV * eV_a_J

# Ecuaciones para los vectores de onda k1 y k2
k1 = mt.sqrt(2 * m * E) / hbar
k2 = mt.sqrt(2 * m * (E - V)) / hbar

# Probabilidades de Transmisión (T) y Reflexión (R)
T = (4 * k1 * k2) / (k1 + k2)**2
R = ((k1 - k2) / (k1 + k2))**2

print("Probabilidad de Transmisión: ", T)
print("Probabilidad de Reflexión: ", R)
```
